```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file = "S:/employee_resignation/Employee_Resignation_ProjectDataset111.xlsx"

df = pd.read_excel(file, sheet_name="Cleaned_Data")

df.head()
```

```python
# 1. Missing values
df.isnull().sum()
```

```python
# 2. Duplicate rows
df.duplicated().sum()
```

```python
# 3. Duplicate Employee IDs
df['EmployeeID'].duplicated().sum()
```

```python
# 4. Invalid dates
df[df['DateOfExit'] < df['DateOfJoining']]
```

```python

# 5. Invalid performance ratings
df[~df['PerformanceRating'].between(1, 5)]
```

```python
#phase: 3 : 
#Data Preprocessing & Feature Engineering
# Calculate tenure
df['CalculatedTenure'] = (
    df['DateOfExit'] - df['DateOfJoining']
).dt.days / 365

df['CalculatedTenure']=df['CalculatedTenure'].round(2)
# Extract exit year and month
df['CalculatedYear'] = df['DateOfExit'].dt.year
df['CalculatedMonth'] = df['DateOfExit'].dt.month_name()
# Compare existing and calculated values
df[['Tenure(Years)', 'CalculatedTenure',
    'ExitYear', 'CalculatedYear',
    'ExitMonth', 'CalculatedMonth']].head(10)
```

```python
#overall summary statistics :
print("Total Resignations:", len(df))

print("Average Age:", df['Age'].mean())

print("Average Tenure:", df['Tenure(Years)'].mean())

print("Average Performance Rating:",
      df['PerformanceRating'].mean())

print("\nGender Distribution:")
print(df['Gender'].value_counts())

print("\nDepartment Distribution:")
print(df['Department'].value_counts())

print("\nResignation Reasons:")
print(df['ResignationReason'].value_counts())
```

```python
#department wise resignation analysis :
dept_analysis = df.groupby('Department').agg(
    Total_Resignations=('EmployeeID', 'count'),
    Average_Age=('Age', 'mean'),
    Average_Tenure=('Tenure(Years)', 'mean'),
    Average_Rating=('PerformanceRating', 'mean')
).round(2).sort_values(
    by='Total_Resignations', ascending=False
)

print(dept_analysis)
```

```python
# <!-- Departmental analysis -->
print("==============================")
print("    departmental analysis")
print("==============================")
dept_percentage = (
    df['Department'].value_counts(normalize=True) * 100
).round(2)

print(dept_percentage)
```

```python
#department bar chart : 
dept_counts = df['Department'].value_counts()

plt.figure(figsize=(10, 5))

sns.barplot(
    x=dept_counts.index,
    y=dept_counts.values
)

plt.title("Department-wise Resignations")
plt.xlabel("Department")
plt.ylabel("Number of Resignations")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Resignation reasons : 
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x='ResignationReason',
    order=df['ResignationReason'].value_counts().index
)

plt.title("Resignations by Reason")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

```python
#Employee Demographic Analysis: 
#1.AGe group analysis: 
df.groupby('AgeGroup').agg(
    Total_Resignations=('EmployeeID', 'count'),
    Average_Performance=('PerformanceRating', 'mean')
).round(2)
```

```python
#performace rating analysis: 
df.groupby('PerformanceRating').agg(
    Total_Resignations=('EmployeeID', 'count'),
    Average_Age=('Age', 'mean'),
    Average_Tenure=('Tenure(Years)', 'mean')
).round(2)
```

```python
#Tenure band analysis: 
df.groupby('TenureBand').agg(
    Total_Resignations=('EmployeeID', 'count'),
    Average_Age=('Age', 'mean'),
    Average_Rating=('PerformanceRating', 'mean')
).round(2)
```

```python
#yearly trends: 
yearly = df.groupby('ExitYear').size()

yearly.plot(
    kind='line',
    marker='o',
    figsize=(10, 5)
)

plt.title("Yearly Resignation Trends")
plt.xlabel("Year")
plt.ylabel("Resignations")
plt.grid()
plt.show()
```

```python
#monthly trends : 


monthly = df.groupby('ExitMonth').size()

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly = monthly.reindex(month_order)

monthly.plot(kind='bar', figsize=(10, 5))

plt.title("Monthly Resignation Trends")
plt.xlabel("Month")
plt.ylabel("Resignations")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#resignation reason by department : 
pivot = pd.crosstab(
    df['Department'],
    df['ResignationReason']
)

sns.heatmap(pivot, annot=True, fmt='d', cmap='YlGnBu')

plt.title("Department vs Resignation Reason")
plt.tight_layout()
plt.show()
```

```python
#resignation reason analysis: 
reason_analysis = df.groupby('ResignationReason').agg(
    Total=('EmployeeID', 'count'),
    Average_Age=('Age', 'mean'),
    Average_Tenure=('Tenure(Years)', 'mean'),
    Average_Rating=('PerformanceRating', 'mean')
).round(2).sort_values('Total', ascending=False)

print(reason_analysis)
```

Final Visualizations

```python
#performance rating distribution: 
sns.countplot(data=df, x='PerformanceRating')
plt.title('Resignations by Performance Rating')
plt.show()
```

```python
#Age vs Tenure: 

plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='Age',
    y='Tenure(Years)',
    hue='Gender',
    alpha=0.5
)

plt.title('Age vs Tenure of Resigned Employees')
plt.show()
```

```python
#resignation readon distrubution:
plt.figure(figsize=(10, 5))

df['ResignationReason'].value_counts().plot(
    kind='bar',
    color='steelblue'
)

plt.title('Resignation Reasons')
plt.ylabel('Number of Resignations')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Export Excel Report
summary = pd.DataFrame({
    'Metric': [
        'Total Resignations',
        'Average Age',
        'Average Tenure',
        'Average Performance Rating'
    ],
    'Value': [
        len(df),
        round(df['Age'].mean(), 2),
        round(df['Tenure(Years)'].mean(), 2),
        round(df['PerformanceRating'].mean(), 2)
    ]
})

department_summary = df.groupby('Department').agg(
    Total_Resignations=('EmployeeID', 'count'),
    Average_Age=('Age', 'mean'),
    Average_Tenure=('Tenure(Years)', 'mean'),
    Average_Rating=('PerformanceRating', 'mean')
).round(2)

reason_summary = df['ResignationReason'].value_counts().reset_index()

with pd.ExcelWriter('Employee_Resignation_Report.xlsx',
                    engine='openpyxl') as writer:

    summary.to_excel(writer, sheet_name='Overall Summary', index=False)

    department_summary.to_excel(
        writer, sheet_name='Department Analysis'
    )

    reason_summary.to_excel(
        writer, sheet_name='Resignation Reasons', index=False
    )

print("Report exported successfully!")
```

```python
# Export everything into one folder
import os

os.makedirs('analysis_results', exist_ok=True)

summary.to_csv('analysis_results/overall_summary.csv', index=False)

department_summary.to_csv(
    'analysis_results/department_analysis.csv'
)

reason_summary.to_csv(
    'analysis_results/resignation_reasons.csv', index=False
)

yearly.to_csv(
    'analysis_results/yearly_trend.csv',
    header=['Total_Resignations']
)

monthly.to_csv(
    'analysis_results/monthly_trend.csv',
    header=['Total_Resignations']
)

pivot.to_csv(
    'analysis_results/department_reason_heatmap_data.csv'
)

print("All analysis results exported successfully!")
```
